```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
```

```python
df = pd.read_csv('height_weight.csv')
```

```python
df.head()
```

```python
## Scatter Plot
plt.scatter(df['Weight'], df['Height'])
plt.title('Weight vs Height')
plt.xlabel('Weight')
plt.ylabel('Height')
```

```python
## Correlation
df.corr()
```

```python
## seaborn library for better visualization
import seaborn as sns
sns.pairplot(df)
```

```python
## Independent and Dependent Variables
X = df[['Weight']] ## Independent features should be dataframe or 2d
print(type(X))
print(type(df))
np.array(X).shape   
```

```python
## Independent and Dependent Variables
X_ = df['Weight'] ## see what happens when we use series instead of dataframe
print(type(X_))
print(type(df))
np.array(X_).shape
```

```python
## Dependent variable

y = df['Height'] ## this variable can be in series or in 1d array
np.array(y).shape
```

```python
## Train Test Split
from sklearn.model_selection import train_test_split

X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.25,random_state=42)
```

```python
X_train.shape
```

```python
## Standardization very very important
### apply the standardization on all independent feature to reduce the convergence
from sklearn.preprocessing import StandardScaler
```

```python
scaler = StandardScaler()
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

```python
X_test
```

```python
## Apply simple linear Regression 
from sklearn.linear_model import LinearRegression
regression = LinearRegression()

```

```python
regression.fit(X_train,y_train) ## it always expect the 2d array
```

```python
regression.coef_
```

```python
regression.intercept_
```

```python
## plot the training data and the regression line
plt.scatter(X_train,y_train)
plt.plot(X_train,regression.predict(X_train))
```

### Prediction of test data
predicted height output = intercept + weight * coefficient

y_pred_test = 156.47 + 17.29(X_test)

```python
## Prediction for the test data
y_pred = regression.predict(X_test)
```

```python
## Performance Metrices
from sklearn.metrics import mean_squared_error, mean_absolute_error
mse  = mean_squared_error(y_test,y_pred)
mae = mean_absolute_error(y_test,y_pred)
rmse = np.sqrt(mse)

print(mse)
print(mae)
print(rmse)
```

```python
from sklearn.metrics import r2_score
score = r2_score(y_test,y_pred)
adjusted_score = 1-(1-score)*(len(y_test)-1)/((len(y_test)-X_test.shape[1]-1))
print(score)
print(adjusted_score)
```
