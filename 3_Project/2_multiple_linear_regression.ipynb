```python
## lets prepare the dataframe
dataset = pd.DataFrame(california.data, columns=california.feature_names)
dataset.head()
```

```python
dataset['Price'] = california.target
```

```python
## Independent and dependent Features
dataset.head()
```

```python
X = dataset.iloc[:,:-1]
y = dataset.iloc[:,-1]

```

```python
X.head()
```

```python
from sklearn.model_selection import train_test_split
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.33,random_state=10)
```

```python
X_train.shape,X_test.shape,y_train.shape,y_test.shape
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
X_train = scaler.fit_transform(X_train)
X_test = scaler.transform(X_test)
```

## Model Training

```python
from sklearn.linear_model import LinearRegression
regressor = LinearRegression()
```

```python
regressor.fit(X_train,y_train)
```

```python
## Prediction for the test data
y_pred = regressor.predict(X_test)
y_pred
```

```python
## Performance Metrices
from sklearn.metrics import mean_squared_error, mean_absolute_error
mse = mean_squared_error(y_test,y_pred)
mae = mean_absolute_error(y_test,y_pred)
rmse = np.sqrt(mse)

print(mse)
print(mae)
print(rmse)
```

```python
## r square and adjusted r square
from sklearn.metrics import r2_score

score = r2_score(y_test,y_pred)

print(score)
```

```python
adjusted_score = 1-(1-score)*(len(y_test)-1)/(len(y_test)-X_train.shape[1]-1)
print(adjusted_score)
```

## Pickling Concept

Python pickle module is used for serializing and de-serializing a python object structure. Any object in python can be pickled so that it can be saved on disk.

```python
import pickle as pkl
pkl.dump(regressor,open('regressor.pkl','wb'))
```

```python
## to load the pickle file
model = pkl.load(open('regressor.pkl','rb'))
```

```python
model.predict(X_test)
```
